```python

# ============================================================
# ACTIVIDAD 5 - PREPARACIÓN DEL ENTRENAMIENTO
# ============================================================

import os
import pandas as pd
import numpy as np
import mlflow

from sklearn.model_selection import train_test_split

# Ruta del dataset limpio
ruta_datos = "/content/Actividad5/datos/datos_limp/datos_limpios.csv"

# Comprobar que el archivo exista
assert os.path.exists(ruta_datos), (
    "No se encontró el CSV limpio. Revisa la carpeta del proyecto."
)

# Cargar los datos
df = pd.read_csv(ruta_datos)

# Separar variables predictoras y objetivo
X = df.drop(columns=["diagnostico"])

# En nuestro dataset:
# 0 = maligno
# 1 = benigno
#
# Para facilitar la interpretación de las métricas,
# transformamos la clase maligna en la clase positiva (1).
y = (df["diagnostico"] == 0).astype(int)

# Separar entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Datos preparados para el entrenamiento")
print("Registros de entrenamiento:", len(X_train))
print("Registros de prueba:", len(X_test))
print("Clase positiva (1): diagnóstico maligno")
print("Casos malignos en entrenamiento:", int(y_train.sum()))
print("Casos malignos en prueba:", int(y_test.sum()))
```

```python
# ============================================================
# CREACIÓN DEL SCRIPT train.py
# ============================================================

codigo_train = r'''
import os
# Permitir el almacenamiento de datos en el sistema de archivos local de MLflow
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

import json
import mlflow
import mlflow.sklearn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, ConfusionMatrixDisplay, RocCurveDisplay
)

def entrenar(ruta_datos, carpeta_salida):
    os.makedirs(carpeta_salida, exist_ok=True)

    df = pd.read_csv(ruta_datos)
    X = df.drop(columns=["diagnostico"])
    y = (df["diagnostico"] == 0).astype(int)

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.20, random_state=42, stratify=y
    )

    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    # El escalado se ajusta dentro de cada partición.
    modelos = {
        "Regresion_Logistica": (
            Pipeline([
                ("escalador", StandardScaler()),
                ("modelo", LogisticRegression(max_iter=5000, random_state=42))
            ]),
            {
                "modelo__C": [0.1, 1, 10],
                "modelo__class_weight": [None, "balanced"]
            }
        ),
        "Random_Forest": (
            Pipeline([
                ("modelo", RandomForestClassifier(random_state=42, n_jobs=1))
            ]),
            {
                "modelo__n_estimators": [100, 200],
                "modelo__max_depth": [None, 10],
                "modelo__min_samples_leaf": [1, 2]
            }
        )
    }

    mlflow.set_tracking_uri("file:/content/Actividad5/mlruns")
    mlflow.set_experiment("Actividad5_Comparacion_Modelos")

    resultados = []

    for nombre, (pipeline, rejilla) in modelos.items():
        busqueda = GridSearchCV(
            estimator=pipeline,
            param_grid=rejilla,
            scoring="f1",
            cv=cv,
            n_jobs=-1,
            refit=True,
            return_train_score=False
        )

        with mlflow.start_run(run_name=nombre):
            busqueda.fit(X_train, y_train)
            mejor_modelo = busqueda.best_estimator_

            predicciones = mejor_modelo.predict(X_test)
            probabilidades = mejor_modelo.predict_proba(X_test)[:, 1]

            metricas = {
                "accuracy": accuracy_score(y_test, predicciones),
                "precision": precision_score(y_test, predicciones, zero_division=0),
                "recall": recall_score(y_test, predicciones, zero_division=0),
                "f1": f1_score(y_test, predicciones, zero_division=0),
                "roc_auc": roc_auc_score(y_test, probabilidades),
                "f1_cv_mejor": busqueda.best_score_
            }

            mlflow.log_params(busqueda.best_params_)
            mlflow.log_param("algoritmo", nombre)
            mlflow.log_param("validacion_cruzada", "StratifiedKFold_5")
            mlflow.log_param("semilla", 42)
            mlflow.log_metrics(metricas)

            # Registrar el modelo entrenado como artefacto permitiendo tipos de árboles seguros
            mlflow.sklearn.log_model(
                mejor_modelo,
                artifact_path="modelo",
                skops_trusted_types=["sklearn.tree._tree.Tree"]
            )

            # Guardar matriz de confusión
            fig, ax = plt.subplots(figsize=(5, 4))
            ConfusionMatrixDisplay.from_predictions(
                y_test, predicciones,
                display_labels=["Benigno", "Maligno"],
                ax=ax
            )
            ax.set_title(f"Matriz de confusión: {nombre}")
            fig.tight_layout()

            ruta_cm = os.path.join(carpeta_salida, f"{nombre}_confusion.png")
            fig.savefig(ruta_cm, dpi=150)
            plt.close(fig)
            mlflow.log_artifact(ruta_cm)

            # Guardar curva ROC
            fig, ax = plt.subplots(figsize=(5, 4))
            RocCurveDisplay.from_predictions(y_test, probabilidades, ax=ax)
            ax.set_title(f"Curva ROC: {nombre}")
            fig.tight_layout()

            ruta_roc = os.path.join(carpeta_salida, f"{nombre}_roc.png")
            fig.savefig(ruta_roc, dpi=150)
            plt.close(fig)
            mlflow.log_artifact(ruta_roc)

            resultados.append({
                "modelo": nombre,
                **metricas,
                "mejores_parametros": json.dumps(
                    busqueda.best_params_, default=str
                )
            })

            print(f"Modelo: {nombre}")
            print("Mejores parámetros:", busqueda.best_params_)
            print("Métricas:", metricas)
            print("-" * 50)

    tabla = pd.DataFrame(resultados)
    ruta_resultados = os.path.join(carpeta_salida, "comparacion_modelos.csv")
    tabla.to_csv(ruta_resultados, index=False)

    return tabla

if __name__ == "__main__":
    tabla = entrenar(
        "/content/Actividad5/datos/datos_limp/datos_limpios.csv",
        "/content/Actividad5/evidencias/graficas"
    )
    print("\nCOMPARACIÓN FINAL")
    print(tabla.to_string(index=False))
'''

ruta_train = "/content/Actividad5/fuentes/train.py"

with open(ruta_train, "w", encoding="utf-8") as archivo:
    archivo.write(codigo_train.strip() + "\n")

print("Archivo train.py creado correctamente con el fix de MLflow y Skops")
print(ruta_train)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import mlflow
from mlflow.tracking import MlflowClient

# --------------------------------------------------
# 1. Configurar MLflow y la carpeta de evidencias
# --------------------------------------------------
mlflow.set_tracking_uri("file:/content/Actividad5/mlruns")
cliente = MlflowClient()

carpeta = "/content/Actividad5/evidencias/mlflow"
os.makedirs(carpeta, exist_ok=True)

experimento = cliente.get_experiment_by_name(
    "Actividad5_Comparacion_Modelos"
)

corridas = cliente.search_runs(
    experiment_ids=[experimento.experiment_id],
    max_results=100
)

# --------------------------------------------------
# 2. Recopilar información de corridas finalizadas
# --------------------------------------------------
filas = []

for corrida in corridas:
    if corrida.info.status != "FINISHED":
        continue

    fila = {
        "modelo": corrida.data.tags.get("mlflow.runName"),
        "estado": corrida.info.status,
        "id_corrida": corrida.info.run_id
    }

    # Parámetros
    for nombre, valor in corrida.data.params.items():
        fila[f"param_{nombre}"] = valor

    # Métricas
    for nombre, valor in corrida.data.metrics.items():
        fila[f"metrica_{nombre}"] = valor

    filas.append(fila)

df_mlflow = pd.DataFrame(filas)

# Guardar el detalle completo de las corridas
ruta_detalle = os.path.join(carpeta, "corridas_mlflow.csv")
df_mlflow.to_csv(ruta_detalle, index=False, encoding="utf-8-sig")

print("Guardado:", ruta_detalle)
display(df_mlflow)

# --------------------------------------------------
# 3. Crear inventario de artefactos por corrida
# --------------------------------------------------
ruta_inventario = os.path.join(
    carpeta, "inventario_artefactos.txt"
)

with open(ruta_inventario, "w", encoding="utf-8") as archivo:
    archivo.write("INVENTARIO DE ARTEFACTOS DE MLFLOW\n")
    archivo.write(f"Experimento: {experimento.name}\n")
    archivo.write(f"ID del experimento: {experimento.experiment_id}\n")

    for corrida in corridas:
        if corrida.info.status != "FINISHED":
            continue

        nombre = corrida.data.tags.get("mlflow.runName", "Sin nombre")
        archivo.write("\n" + "=" * 50 + "\n")
        archivo.write(f"Modelo: {nombre}\n")
        archivo.write(f"Estado: {corrida.info.status}\n")
        archivo.write(f"ID: {corrida.info.run_id}\n")

        for artefacto in cliente.list_artifacts(corrida.info.run_id):
            archivo.write(f"Artefacto: {artefacto.path}\n")

print("Guardado:", ruta_inventario)

# --------------------------------------------------
# 4. Crear tabla resumen de métricas
#    Una sola fila por modelo
# --------------------------------------------------
columnas_metricas = [
    "metrica_accuracy",
    "metrica_precision",
    "metrica_recall",
    "metrica_f1",
    "metrica_roc_auc",
    "metrica_f1_cv_mejor"
]

df_resumen = df_mlflow.drop_duplicates(
    subset=["modelo"],
    keep="first"
).copy()

columnas_disponibles = [
    c for c in columnas_metricas if c in df_resumen.columns
]

tabla = df_resumen[["modelo"] + columnas_disponibles].copy()

# Mostrar porcentajes de forma legible
for columna in columnas_disponibles:
    tabla[columna] = (tabla[columna] * 100).round(2)

tabla = tabla.rename(columns={
    "modelo": "Modelo",
    "metrica_accuracy": "Accuracy (%)",
    "metrica_precision": "Precision (%)",
    "metrica_recall": "Recall (%)",
    "metrica_f1": "F1 (%)",
    "metrica_roc_auc": "AUC-ROC (%)",
    "metrica_f1_cv_mejor": "F1 CV (%)"
})

ruta_resumen = os.path.join(carpeta, "resumen_metricas_mlflow.csv")
tabla.to_csv(ruta_resumen, index=False, encoding="utf-8-sig")

print("Guardado:", ruta_resumen)
display(tabla)

# --------------------------------------------------
# 5. Crear imagen de la tabla de métricas
# --------------------------------------------------
fig, ax = plt.subplots(figsize=(12, 2.5))
ax.axis("off")

tabla_grafica = ax.table(
    cellText=tabla.values,
    colLabels=tabla.columns,
    loc="center",
    cellLoc="center"
)

tabla_grafica.auto_set_font_size(False)
tabla_grafica.set_fontsize(8)
tabla_grafica.scale(1, 1.7)

ax.set_title(
    "MLflow: comparación de corridas finalizadas",
    fontsize=13,
    pad=18
)

ruta_png = os.path.join(carpeta, "resumen_metricas_mlflow.png")
plt.savefig(ruta_png, dpi=180, bbox_inches="tight")
plt.show()
plt.close(fig)

print("\nEvidencias creadas:")
for nombre in sorted(os.listdir(carpeta)):
    print("-", nombre)
```
